```python
import pandas as pd

df = pd.read_csv("applicants.csv")
```

```python
Markdown (Observation):
Dataset loaded successfully. It contains applicant records with 7 columns (Name, Email, Phone, Domain Applied, University, Application Date, Status).
```

```python
Markdown (Observation):

Missing values found in Applicant Name, Email, University, and Status.

Some duplicate rows exist (e.g., Abdullah Chaudhry, Hassan Hussain, Noor Ali).

Application Date stored as text with mixed formats.

Phone numbers inconsistent with spaces/dashes.
```

```python
# Drop rows with missing Applicant Name
df = df.dropna(subset=['Applicant Name'])
```

```python
# Fill missing Email with placeholder
df['Email'] = df['Email'].fillna("not_provided@example.com")
```

```python
# Fill missing University
df['University'] = df['University'].fillna("Not Specified")
```

```python
# Fill missing Status
df['Status'] = df['Status'].fillna("Under Review")
```

```python
Markdown (Observation):

Applicant Name missing → unusable rows.

Email missing → critical for communication.

University missing → non‑critical.

Status missing → default assumed "Under Review".
```

```python
Markdown (After Fix):

Dropped nameless rows.

Filled missing Email with placeholder.

Filled missing University with "Not Specified".

Filled missing Status with "Under Review".
```

```python
# Remove exact duplicates
df = df.drop_duplicates()
```

```python
# Remove duplicates by Email
df = df.drop_duplicates(subset=['Email'], keep='first')
```

```python
Markdown (Observation):

Found duplicate applicants (same Email repeated).

Markdown (After Fix):

Removed exact duplicates and kept only one record per Email.
```

```python
# Strip whitespace
df['Applicant Name'] = df['Applicant Name'].str.strip()
```

```python
# Standardize Status casing
df['Status'] = df['Status'].str.strip().str.title()
```

```python
# Domain mapping
domain_mapping = {
    'web dev':'Web Development',
    'webdev':'Web Development',
    'appdev':'App Development',
    'app dev':'App Development',
    'mobile app development':'App Development',
    'graphics design':'Graphic Design',
    'graphic design':'Graphic Design',
    'digital marketing':'Digital Marketing',
    'digital marketting':'Digital Marketing',
    'datascience':'Data Science',
    'data science':'Data Science'
}
df['Domain Applied'] = df['Domain Applied'].str.strip().str.lower().map(domain_mapping)
```

```python
Markdown (Observation):

Domain Applied column had inconsistent casing/spelling (e.g., "webdev", "WEB DEV", "Digital marketting").

Status column had inconsistent casing ("selected", "SELECTED", "Rejected").
```

```python
Markdown (After Fix):

All Domain Applied values standardized into 5 categories.

Status column standardized to Title Case.
```

```python
# Convert Application Date
df['Application Date'] = pd.to_datetime(df['Application Date'], errors='coerce')
```

```python
# Clean Phone numbers
df['Phone'] = df['Phone'].str.replace('-', '', regex=False).str.replace(' ', '', regex=False)
```

```python
Markdown (Observation):

Application Date stored as text with mixed formats, some invalid (e.g., "not-a-date", "31/19/2026").

Phone numbers inconsistent with spaces/dashes.

Markdown (After Fix):

Application Date converted to proper datetime format; invalid dates marked as NaT.

Phone numbers cleaned to consistent digit strings.
```

```python
Markdown (Final Summary):

Starting dataset: 85 rows, 7 columns।

Issues: Missing values (Name, Email, University, Status), duplicates, inconsistent Domain & Status, mixed date formats, messy phone numbers।

Actions: Dropped nameless rows, filled missing values, removed duplicates, standardized text fields, converted dates, cleaned phone numbers।

Final dataset: Clean, consistent, no duplicates, proper datetime type, standardized categories।
```
